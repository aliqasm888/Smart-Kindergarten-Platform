```python
# !pip install -q flask openai-whisper ffmpeg-python pyngrok
# !pip install sentence-transformers scikit-learn
# !ngrok config add-authtoken 30N1SxkTtjd24HjvcbydnONHSFx_5s9SJ75zbu4cF4bH344Pc

# from sklearn.metrics.pairwise import cosine_similarity
# from flask import Flask, request, jsonify
# from pyngrok import ngrok
# import whisper
# import os
# from sentence_transformers import SentenceTransformer
# import numpy as np
# from sklearn.metrics.pairwise import cosine_similarity

# # تحميل نموذج Whisper
# model = whisper.load_model("base")

# # إعداد Flask
# app = Flask(__name__)

# @app.route('/transcribe', methods=['POST'])
# def transcribe():
#     if 'audio' not in request.files:
#         return jsonify({"error": "يرجى إرسال ملف صوتي بالحقل audio"}), 400

#     audio = request.files['audio']
#     audio.save("temp.mp3")

#     result = model.transcribe("temp.mp3", language="ar")
#     os.remove("temp.mp3")

#     return jsonify({"text": result["text"]})


# SentenceTransformer_model = SentenceTransformer('sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2')

# EXAMPLES = {
#       "بالتأكيد لا": [
#           "أبداً ما بيصير",
#           "مستحيل يعمل هيك",
#           "ولا مرة صارت",
#           "مطلقاً ما بيعمل هيك",
#           "ما صار ولا يوم",
#           "أكيد لأ",
#           "ما بيعملها أبداً",
#           "ما عندو هالشي نهائياً",
#           "محال يسوي هيك",
#           "ما بصير منو هيك سلوك",
#            "مستحيل","اكيد لا","أكيد لا","ابداً","ابدا","نهائياً لا","مو ممكن","لا طبعاً",
#         "شلون اي","ما بيصير ابداً","مرفوض","غير موجود اطلاقاً"  ,"لا "
#       ],

#         "شبه مؤكد لا": [
#         "على الغالب لا",
#         "ما بتوقع",
#         "ما بظن",
#         "نادر كتير",
#         "احتمال ضعيف",
#         "ما بتخيل يصير",
#         "يمكن مرة بس تقريباً لأ",
#         "غالباً ما بيصير",
#         "صراحة ما بيعملها",
#           "شبه اكيد لا","تقريباً لا مؤكد","٩٠% لا","تسعين بالمية لا","اكتر شي لا","غالباً لا اكيد"
#     ],

#   "على الأغلب لا": [
#     "ما كتير بيعملها",
#     "مرة بالألف",
#     "نادر يصير",
#     "ما بصير إلا قليل",
#     "ما بيسويها إلا نادراً",
#     "إلها حالات قليلة",
#     "بيصير شي مرة كل زمان",
#     "على الأغلب لأ",
#        "على الغالب لا","غالب الظن لا","ما اظن","ما أظن","ارجح لا","احتمال كبير لا","غالباً لا"
# ],
# "ربما لا": [
#     "يمكن ما",
#     "أحياناً ما",
#     "ممكن لا",
#     "ما متأكد بس الأغلب لأ",
#     "مو دائماً بيصير",
#     "احتمال قليل لا",
#     "أوقات لا",
#     "ما بعرف يمكن ما",
#     "يمكن مرات ما بيصير",
#       "يمكن لا","بركي لا","مو كتير","مش كتير","ماني راجح","غالباً لا بس مو اكيد",
#         "احتمال خفيف","ضعيف نعم يعني لا","يمكن لا اكتر"
# ],

#    "غير معروف": [
#     "ما بعرف",
#     "صراحة مو متأكد",
#     "ما فيني جاوب",
#     "ما عندي فكرة",
#     "يمكن هيك ويمكن لأ",
#     "محتاج فكر",
#     "ما عندي جواب دقيق",
#      "ما بعرف","ماني متأكد","مش متأكد","مو متأكد","مو واضح","غير واضح","ما عندي فكرة",
#         "ما بقدر احكم","بحاجه معلومات اكتر","بدّي فكر","بدي فكر","لا ادري"
# ]
# ,
#    "ربما": [
#     "يمكن",
#     "ممكن",
#     "أحياناً",
#     "على حسب",
#     "مرات بيصير",
#     "أوقات بيصير",
#     "ما دايماً",
#     "ممكن يكون",
#     "مش دايماً بس بيصير",
#        "يمكن","ممكن","احتمال","بركي","على حسب","نوعاً ما","الى حد ما","لحد ما","شو بعرفني",
#         "ماني متأكد بس يمكن","يمكن اي"
# ]
# ,
#     "على الأغلب": [
#     "غالباً",
#     "بيصير كتير",
#     "معظم الوقت بيعملها",
#     "بيسويها عادة",
#     "بيصير معه أكتر من مرة",
#     "بيصير مرات كتيرة",
#     "بكثرة",
#     "غالباً بيصير",
#     "على الارجح","على الغالب","غالب الظن","احتمال كبير نعم","غالباً نعم","غالبا نعم",
#         "على الغالب نعم","ارجح نعم","بيتهيألي نعم"
# ]
# ,
#     "شبه مؤكد نعم": [
#     "تقريباً دايماً",
#     "أغلب الوقت",
#     "بنسبة كبيرة",
#     "معظم الأحيان",
#     "دائماً تقريباً",
#     "كتير مرات",
#     "كل فترة بيصير",
#     "أكتر الوقت هيك",
#     "غالباً نعم",
#      "شبه اكيد","تقريباً اكيد","تسعين بالمية نعم","٩٠% نعم","على الاغلب اي","على الغالب اي",
#         "اي على ما اظن","اي يمكن اكيد","يبدو نعم","محتمل بقوة","راجح نعم" , "اي"," نعم"
# ]
# ,
#  "بالتأكيد نعم": [
#     "أكيد",
#     "طبعاً",
#     "دائماً",
#     "كل الوقت",
#     "100% بيصير",
#     "ما في شك",
#     "على طول هيك",
#     "ضروري بيعمل هيك",
#     "بلا نقاش نعم",
#     "دائماً يتجنب الناس",
#     "ما بيقرب على حدا أبداً",
#     "ما بيحكي مع حدا مطلقاً",
#     "بيتجنب التواصل طول الوقت",
#     "اكيد نعم","أكيد نعم","اي والله","اي اكيد","طبعاً","مية بالمية","مية بالميه","مؤكد",
#     "بدون شك","تمام التمام","اي اي","اي صحيح","مزبوط","مزبووط","مزبوط كتير",
#     "اي مية بالمية","تمام","اي تمام","صحيح","اي ","أكيد "
# ]

# }
# category_embeddings = {}
# for label, sentences in EXAMPLES.items():
#     sentence_embeddings = model.encode(sentences)
#     average_embedding = np.mean(sentence_embeddings, axis=0)
#     category_embeddings[label] = average_embedding
# print("شكل embedding لـ 'ربما':", category_embeddings["ربما"].shape)
# print(category_embeddings["بالتأكيد نعم"][:10])


# # فتح نفق ngrok للمنفذ 5000
# public_url = ngrok.connect(5000)
# print("🔗 رابط API الخارجي:", public_url)
# @app.route('/classify', methods=['POST'])
# def classify():
#     data = request.json
#     user_text = data.get('text')
#     if not user_text:
#         return jsonify({'error': 'Missing text'}), 400

#     user_embedding = model.encode([user_text])
#     similarities = {}
#     for label, avg_emb in category_embeddings.items():
#         sim = cosine_similarity(user_embedding, [avg_emb])[0][0]
#         similarities[label] = sim

#     sorted_similarities = sorted(similarities.items(), key=lambda x: x[1], reverse=True)
#     best_label = sorted_similarities[0][0]

#     # اطبع التصنيف في التيرمينال
#     print(f"User text: {user_text} => Classified as: {best_label}")

#     return jsonify({'label': best_label})

# user_input = "ابني دائمًا يتجنب الناس"

# user_embedding = model.encode([user_input])

# similarities = {}
# for label, avg_embedding in category_embeddings.items():
#     sim = cosine_similarity(user_embedding, [avg_embedding])[0][0]
#     similarities[label] = sim

# sorted_similarities = sorted(similarities.items(), key=lambda x: x[1], reverse=True)

# print("\nأقرب التصنيفات:")
# for label, score in sorted_similarities:
#     print(f"{label}: {score:.2f}")
# # تشغيل السيرفر
# app.run(port=5000)
# ==============================
# 📌 تنصيب المكتبات (مرة واحدة)
# ==============================
!pip install -q flask openai-whisper ffmpeg-python pyngrok
!pip install sentence-transformers scikit-learn
!ngrok config add-authtoken 30N1SxkTtjd24HjvcbydnONHSFx_5s9SJ75zbu4cF4bH344Pc
# ==============================
# 📌 الاستيراد
# ==============================
from flask import Flask, request, jsonify
from pyngrok import ngrok
import whisper
import os
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# ==============================
# 📌 تحميل نموذج Whisper للتفريغ
# ==============================
whisper_model = whisper.load_model("base")

# ==============================
# 📌 إعداد Flask
# ==============================
app = Flask(__name__)

# ==============================
# 📌 نقطة النهاية: تحويل الصوت إلى نص
# ==============================
@app.route('/transcribe', methods=['POST'])
def transcribe():
    if 'audio' not in request.files:
        return jsonify({"error": "يرجى إرسال ملف صوتي بالحقل audio"}), 400

    audio = request.files['audio']
    audio.save("temp.mp3")

    result = whisper_model.transcribe("temp.mp3", language="ar")
    os.remove("temp.mp3")

    return jsonify({"text": result["text"]})

# ==============================
# 📌 تحميل SentenceTransformer للتصنيف
# ==============================
st_model = SentenceTransformer('sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2')

# ==============================
# 📌 الأمثلة
# ==============================
EXAMPLES = {
    "بالتأكيد لا": [
        "أبداً ما بيصير","مستحيل يعمل هيك","ولا مرة صارت","مطلقاً ما بيعمل هيك","ما صار ولا يوم",
        "أكيد لأ","ما بيعملها أبداً","ما عندو هالشي نهائياً","محال يسوي هيك","ما بصير منو هيك سلوك",
        "مستحيل","اكيد لا","أكيد لا","ابداً","ابدا","نهائياً لا","مو ممكن","لا طبعاً",
        "شلون اي","ما بيصير ابداً","مرفوض","غير موجود اطلاقاً","لا "
    ],
    "شبه مؤكد لا": [
        "على الغالب لا","ما بتوقع","ما بظن","نادر كتير","احتمال ضعيف","ما بتخيل يصير",
        "يمكن مرة بس تقريباً لأ","غالباً ما بيصير","صراحة ما بيعملها",
        "شبه اكيد لا","تقريباً لا مؤكد","٩٠% لا","تسعين بالمية لا","اكتر شي لا","غالباً لا اكيد"
    ],
    "على الأغلب لا": [
        "ما كتير بيعملها","مرة بالألف","نادر يصير","ما بصير إلا قليل","ما بيسويها إلا نادراً",
        "إلها حالات قليلة","بيصير شي مرة كل زمان","على الأغلب لأ",
        "على الغالب لا","غالب الظن لا","ما اظن","ما أظن","ارجح لا","احتمال كبير لا","غالباً لا"
    ],
    "ربما لا": [
        "يمكن ما","أحياناً ما","ممكن لا","ما متأكد بس الأغلب لأ","مو دائماً بيصير",
        "احتمال قليل لا","أوقات لا","ما بعرف يمكن ما","يمكن مرات ما بيصير",
        "يمكن لا","بركي لا","مو كتير","مش كتير","ماني راجح","غالباً لا بس مو اكيد",
        "احتمال خفيف","ضعيف نعم يعني لا","يمكن لا اكتر"
    ],
    "غير معروف": [
        "ما بعرف","صراحة مو متأكد","ما فيني جاوب","ما عندي فكرة","يمكن هيك ويمكن لأ",
        "محتاج فكر","ما عندي جواب دقيق","ما بعرف","ماني متأكد","مش متأكد","مو متأكد","مو واضح",
        "غير واضح","ما عندي فكرة","ما بقدر احكم","بحاجه معلومات اكتر","بدّي فكر","بدي فكر","لا ادري"
    ],
    "ربما": [
        "يمكن","ممكن","أحياناً","على حسب","مرات بيصير","أوقات بيصير","ما دايماً","ممكن يكون",
        "مش دايماً بس بيصير","يمكن","ممكن","احتمال","بركي","على حسب","نوعاً ما","الى حد ما","لحد ما",
        "شو بعرفني","ماني متأكد بس يمكن","يمكن اي"
    ],
    "على الأغلب": [
        "غالباً","بيصير كتير","معظم الوقت بيعملها","بيسويها عادة","بيصير معه أكتر من مرة",
        "بيصير مرات كتيرة","بكثرة","غالباً بيصير","على الارجح","على الغالب","غالب الظن",
        "احتمال كبير نعم","غالباً نعم","غالبا نعم","على الغالب نعم","ارجح نعم","بيتهيألي نعم"
    ],
    "شبه مؤكد نعم": [
        "تقريباً دايماً","أغلب الوقت","بنسبة كبيرة","معظم الأحيان","دائماً تقريباً","كتير مرات",
        "كل فترة بيصير","أكتر الوقت هيك","غالباً نعم","شبه اكيد","تقريباً اكيد","تسعين بالمية نعم",
        "٩٠% نعم","على الاغلب اي","على الغالب اي","اي على ما اظن","اي يمكن اكيد","يبدو نعم",
        "محتمل بقوة","راجح نعم","اي","نعم"
    ],
    "بالتأكيد نعم": [
        "أكيد","طبعاً","دائماً","كل الوقت","100% بيصير","ما في شك","على طول هيك","ضروري بيعمل هيك",
        "بلا نقاش نعم","دائماً يتجنب الناس","ما بيقرب على حدا أبداً","ما بيحكي مع حدا مطلقاً",
        "بيتجنب التواصل طول الوقت","اكيد نعم","أكيد نعم","اي والله","اي اكيد","طبعاً",
        "مية بالمية","مية بالميه","مؤكد","بدون شك","تمام التمام","اي اي","اي صحيح","مزبوط",
        "مزبووط","مزبوط كتير","اي مية بالمية","تمام","اي تمام","صحيح","اي","أكيد "
    ]
}

# ==============================
# 📌 إنشاء Embeddings للفئات
# ==============================
category_embeddings = {}
for label, sentences in EXAMPLES.items():
    sentence_embeddings = st_model.encode(sentences)
    average_embedding = np.mean(sentence_embeddings, axis=0)
    category_embeddings[label] = average_embedding

print("✅ شكل embedding لـ 'ربما':", category_embeddings["ربما"].shape)
print("✅ أول 10 قيم من 'بالتأكيد نعم':", category_embeddings["بالتأكيد نعم"][:10])

# ==============================
# 📌 نقطة النهاية: تصنيف النصوص
# ==============================
@app.route('/classify', methods=['POST'])
def classify():
    data = request.get_json(silent=True) or {}
    user_text = data.get('text')

    if not user_text:
        return jsonify({'error': 'Missing text'}), 400

    # حساب embedding للنص
    user_embedding = st_model.encode([user_text])

    # حساب التشابهات وتحويلها إلى float عادي
    similarities = {}
    for label, avg_emb in category_embeddings.items():
        sim = cosine_similarity(user_embedding, [avg_emb])[0][0]
        similarities[label] = float(sim)  # <- أهم سطر

    # ترتيب النتائج
    sorted_similarities = sorted(similarities.items(), key=lambda x: x[1], reverse=True)
    best_label = sorted_similarities[0][0]

    print(f"User text: {user_text} => Classified as: {best_label}")

    # تجهيز شكل مناسب JSON (قائمة من كائنات) والتأكد أن القيم float عادية
    similarities_json = [
        {'label': lbl, 'score': float(score)} for lbl, score in sorted_similarities
    ]

    return jsonify({'label': best_label, 'similarities': similarities_json})


# print("\n✅ أقرب التصنيفات:")
# for label, score in sorted_similarities:
#     print(f"{label}: {score:.2f}")

# ==============================
# 📌 فتح ngrok وتشغيل السيرفر
# ==============================
if __name__ == "__main__":
    public_url = ngrok.connect(5000)
    print("🔗 رابط API الخارجي:", public_url)
    app.run(port=5000)
```
